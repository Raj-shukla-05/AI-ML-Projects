```python
# STEP 1: Setup
import shap
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import pickle
import os

# Make sure folder exists
os.makedirs("../visuals/explainability", exist_ok=True)

# STEP 2: Load Model & Data
with open("../models/best_model.pkl", "rb") as f:
    model = pickle.load(f)

X = pd.read_csv("../data/diabetes.csv").drop("Outcome", axis=1)

# Clean
cols_to_replace = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
X[cols_to_replace] = X[cols_to_replace].replace(0, np.NaN)
X.fillna(X.median(), inplace=True)

# STEP 3: Explain with SHAP
explainer = shap.Explainer(model)
shap_values = explainer(X)

# Summary Plot
shap.summary_plot(shap_values, X, show=False)
plt.savefig("../visuals/explainability/shap_summary_plot.png", bbox_inches='tight')
plt.close()

# STEP 4: Feature Importance Plot
shap.plots.bar(shap_values, show=False)
plt.title("SHAP Feature Importance")
plt.savefig("../visuals/explainability/shap_feature_importance.png", bbox_inches='tight')
plt.close()
```
